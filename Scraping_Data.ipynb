# Step of Scraping Data

Scraping data review and rating in My UNP app from Google Play Store

```python
!pip install google-play-scraper
```

```python
from google_play_scraper import app, reviews, Sort

# Fetch app details
app_details = app('com.unp.newmyunp')  # Replace with the app's package name
print("App Title:", app_details['title'])
print("Rating:", app_details['score'])
print("Total Reviews:", app_details['reviews'])

# Fetch reviews
result, continuation_token = reviews(
    'com.unp.newmyunp',  # Replace with the app's package name
    lang='id',          # Language (e.g., 'en' for English)
    country='id',       # Country (e.g., 'us' for United States)
    sort=Sort.MOST_RELEVANT,  # Sorting: MOST_RELEVANT, NEWEST, etc.
    count=5000           # Number of reviews to fetch
)

# Print fetched reviews
for review in result:
    print(f"Review by {review['userName']}:")
    print(f"Rating: {review['score']}")
    print(f"Content: {review['content']}")
    print("=" * 100)
```

```python
# konversi hasil scraping ke bentuk dataframe
import pandas as pd

df = pd.DataFrame(result)
df.head()
```

```python
# memilih baris yang akan digunakan untuk analisis
data_review = df[['reviewId', 'content', 'score']]
data_review.head()
```

```python
# save final result ke csv
data_review.to_csv("myunp_reviews.csv", index = False)
```
